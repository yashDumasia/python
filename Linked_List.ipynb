```python
class Node:
    
    def __init__(self,value):
        self.data = value
        self.next = None        
```

```python
class LinkedList:
    
    def __init__(self):
        self.head = None
        self.n = 0
    
    def __len__(self):
        return self.n
    
    def __str__(self):
        
        current = self.head
        result = ""
        
        while current != None:
            result = result + f"{current.data}" + "->"
            current = current.next
            
        return result[:-2]
    
    def __getitem__(self,index):
        
        if index >= self.n:
            return f"{index} is not in linked list"
        
        current = self.head
        x = 0
        
        while x != index:  
            current = current.next
            x = x + 1
            
        return current.data
         
    def insert_head(self,value):
        
        # Create New Node
        new_node = Node(value)    
        
        # Create Connection 
        new_node.next = self.head 
        
        # Reassign Head  
        self.head = new_node
        
        # increment n
        self.n = self.n + 1
        
    def insert_tail(self,value):
        
        new_node = Node(value)    

        if self.head == None:
            
            self.head = new_node
            self.n = self.n + 1
            return
            
        current = self.head
        
        while current.next != None:
            current = current.next
        
        current.next = new_node
        
        self.n = self.n + 1
        
    def insert_after(self,after,value):
        
        new_node = Node(value)
        current = self.head
        x = 0
        
        while current != None:
            
            if current.data == after:
                new_node.next = current.next
                current.next = new_node
                self.n = self.n + 1
                x += 1
                break
            
            current = current.next
            
        if x == 0:
            print(f"There are not {after} in Linked List.")
            
    def clear(self):
        
        self.head = None
        self.n = 0
        
    def delete_head(self):
        
        if self.head == None:
            return "Empty Linked List"
        
        self.head = self.head.next    
        self.n = self.n - 1
        
    def delete_tail(self):
        
        current = self.head
        
        if self.head == None:
            return "Empty Linked List"
        
        elif current.next == None:
            self.delete_head() 
            return

        while current != None:    
            
            if current.next.next == None:
                current.next = None
                self.n = self.n - 1
                break
            
            current = current.next
            
    def remove(self,value):
        
        current = self.head
        x = 1
        
        if self.head == None:
            return "Empty Linked List"
        elif current.data == value:
            self.delete_head()
            return
        
        while current.next != None:
            if current.next.data == value:
                current.next = current.next.next
                self.n = self.n - 1
                break
            current = current.next
            x += 1
            
        if x == self.n:
            return f"{value} is not in Linked List."
        
    def find(self,value): 

        current = self.head
        index = 0
        
        while current != None:
            if current.data == value:
                return f"index : {index}"
            current = current.next
            index += 1
        
        else:
            return f"{value} is not in Linked List."
        
    def replace_max(self,value):
        
        max = self.head
        current = self.head
        
        if current == None:
            return "Empty Linked List"
              
        while current != None:
            
            if current.data > max.data:
                max = current 
            
            current = current.next
        
        max.data = value
        return 
    
    def sum_odd(self):
        
        current = self.head
        x = 0
        sum = 0
        
        if current == None:
            return "Empty Linked List"
        
        while current != None:
            
            if x % 2 == 0:
                sum = sum + current.data 
            
            current = current.next
            x = x + 1 
        
        return sum
    
    def reversed(self):
        
        prev_node = None
        current_node = self.head
        
        while current_node != None:
            
            next_node = current_node.next
            current_node.next = prev_node
            
            prev_node = current_node
            current_node = next_node
            
        self.head = prev_node
        
    def world_LL(self):
        
        current = self.head
        result = ""
        
        while current != None:
            
            if current.data == "/" and current.next.data == "/" or current.data == "*" and current.next.data == "*":
                current.next.next.data = current.next.next.data.upper()
                current.next = current.next.next
            
            if current.data == "*" and current.next.data == "/" or current.data == "/" and current.next.data == "*":
                current.next.next.data = current.next.next.data.upper()
                current.next = current.next.next

            
            if current.data == "/" or current.data == "*":
                current.data = " "
            
            
            result = result + f"{current.data}" 
            current = current.next
        
        return result
```
