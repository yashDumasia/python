```python
class Node:
    
    def __init__(self,value):
        self.data = value
        self.next = None
```

```python
class Queue:
    
    def __init__(self):
        self.front = None
        self.rear = None

    def enqueue(self,value):
        new = Node(value)
        
        if self.rear == None:
            self.front = new
            self.rear = new
            
        else:
            self.rear.next = new
            self.rear = new
            
    def dequeue(self):
        
        if self.front == None:
            return "Empty Queue"
        
        else : 
            x = self.front
            self.front = self.front.next
            print(x.data)
            return 
            
    def traverse(self):
        
        current = self.front
        
        while current != None:
            print(current.data, end=" ")
            current = current.next
            
    def isempty(self):
        
        return self.front == None
    
    def size(self):
        
        current = self.front
        x = 0
        
        while current != None:
            x += 1 
            current = current.next
    
        return x
    
    def front_item(self):
        if self.front == None:
            return "Empty Queue"
        return self.front.data
    
    def rear_item(self):
        if self.front == None:
            return "Empty Queue"
        return self.rear.data 
```

```python
q = Queue()
```

```python
q.front_item()
```

```python
q.enqueue(5)
q.enqueue(10)
q.enqueue(7)
q.enqueue(1)
q.enqueue(12)
q.enqueue(4)
```

```python
q.dequeue()
```

```python
q.traverse()
```

```python
q.isempty()
```

```python
q.size()
```

```python
q.rear_item()
```

```python
class Queue2:
    def __init__(self):
        self.s1 = Stack_LL()
        self.s2 = Stack_LL()
        
    def enqueue(self,value):
        
        self.s1.push(value)
        
    def dequeue(self):
        
        if self.s2.isempty() == False:
            return self.s2.pop()
        else :
            if self.s1.isempty() == True:
                return "Queue is Empty"
            else :
                for i in range(self.s1.size()):
                    self.s2.push(self.s1.pop())
                return self.s2.pop()
            
    def traverse(self):

        # Print s2
        current = self.s2.top
    
        while current != None:
            print(current.data, end=" ")
            current = current.next
    
        # Store s1 elements
        temp = []
        current = self.s1.top
    
        while current != None:
            temp.append(current.data)
            current = current.next
    
        # Print s1 in reverse
        for i in range(len(temp)-1, -1, -1):
            print(temp[i], end=" ")
    
        print()
```
