```python
class Node:
    def __init__(self,value):
        self.data = value
        self.next = None
```

```python
class Stack:
    def __init__(self):
        self.top = None
        self.n = 0
        
    def __str__(self):
        result = ""
        current = self.top
        
        while current != None:
            
            result = result + f"{current.data}" + "\n"
            current = current.next
        
        return result
        
    def isempty(self):
        return self.top == None
    
    def push(self,value):
        New = Node(value)
        
        New.next = self.top
        
        self.top = New
        
        self.n = self.n + 1
        
    def peak(self):
        
        if self.isempty() == True:
            return "Stack is Empty"
        return self.top.data
    
    def pop(self):
        
        if self.isempty() == True:
            return "Stack is Empty"
        
        x = self.top
        self.top = self.top.next
        self.n = self.n - 1
        return x.data
    
    def size(self):
        
        return self.n
    
    def transverse(self):
        
        result = ""
        current = self.top
        
        while current != None:
            
            result = result + f"{current.data}" 
            current = current.next
        
        return result
        
```

```python
def reverse_string(string : str):
    x = Stack()
    y = ""
    
    for i in string:
        x.push(i)
        
    while x.isempty() != True:
        y = y + f"{x.pop()}"
        
    return y
```

```python
def text_editor(string:str,operation:str):
    u = Stack()
    r = Stack()
    v = ""
    
    for i in string:
        u.push(i)
    
    for j in operation:
        
        if j.lower() == "u":
            r.push(u.pop())
        elif j.lower() == "r":
            u.push(r.pop())
        
        if r.isempty() == True:
            break
            
    while u.isempty() != True:
        
        v = u.pop() + v
            
    return v
```

```python
def find_the_celeb(L):
    s = Stack()
    
    for i in range(len(L)):
        s.push(i)
        
    while s.size() >= 2:
        
        i = s.pop()
        j = s.pop()
        
        if L[i][j] == 0:
            s.push(i)
        else:
            s.push(j)
    
    celeb = s.pop()
    
    for i in range(len(L[celeb])):
        if i != celeb:
            if L[celeb][i] == 1 or L[i][celeb] == 0:
                print(None)
                return
    
    print(f"Celeb : {celeb}")  
    return 
```
