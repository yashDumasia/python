```python
import ctypes
```

```python
class Meralist:
    
    def __init__(self):
        self.size = 1 
        self.n = 0 
        # Create a C type array with size = self.size
        self.A = self.__make_array(self.size)
        
    def __len__(self):
        return self.size
    
    def __str__(self):
        result = ''
        for i in range(self.n):
            result = result + str(self.A[i]) + ','
        
        return '[' + result[:-1] + ']'
    
    def __getitem__(self,index):
        if index > self.n :
            print("Index Error.!(Index out of range.!)")
        else:
            return self.A[index]
        
    def __delitem__(self, index):
        if index > self.n:
            print(f"Value Error : {index} index is not in list")
        else:    
            for i in range(index,self.n-1):
                self.A[i] = self.A[i+1]
            self.n = self.n - 1
    
    def append(self,item):
        if self.size == self.n :
           self.resize(self.size*2) # resize
           
        self.A[self.n] = item # append
        self.n = self.n + 1
        
    def resize(self,new_capacity):
        # Create a C type array with new_capacity
        B = self.__make_array(new_capacity)
        self.size = new_capacity
        # Copy content of A to B
        for i in range(self.n):
            B[i] = self.A[i]
            
        self.A = B
        
    def pop(self):
        if self.n == 0:
            return "Empty List"
        else:
            print(self.A[self.n-1])
            self.n = self.n - 1
            
    def clear(self):
        self.size = 1 
        self.n = 0
        
    def find(self,item):
        x = 0
        for i in range(0,self.n):
            if self.A[i] == item:
                print("Index : " + f"{i}")
                x = x + 1
        if x == 0:
            print(f"Value Error : {item} is not in list")
            
    def insert(self,index,item):
        if index > self.n:
            print(f"Value Error : {index} index is not in list")

        if self.n == self.size:
            self.resize(self.n*2)
            
        for i in range(self.n,index,-1):
            self.A[i] = self.A[i-1]
        self.A[index] = item
        self.n = self.n + 1
        
    def remove(self,item):
        x = 0
        for i in range(0,self.n):
            if self.A[i] == item:
                self.__delitem__(i)
                x = x + 1
        if x == 0:
            return f"Value Error : {item} is not in list"
        
    def sort(self):
        x = 0
        for i in range(self.n):
            if isinstance(self.A[i], int):
                x = x+1
        if x == self.n:
            for i in range(self.n-1):
                for j in range(self.n-1-i):
                    if self.A[j] > self.A[j+1]:
                        z = self.A[j]
                        self.A[j] = self.A[j+1]
                        self.A[j+1] = z
        else:    
            return "All Value of List are not Integer so List can not be sort..!!"
        
    def max(self):
        x = 0
        max = self.A[0]

        for i in range(self.n):
            if isinstance(self.A[i], int):
                x = x+1
                
        if x == self.n:
            for i in range(self.n):
                if self.A[i] > max:
                    max = self.A[i]
            return max
        else:    
            return "All Value of List are not Integer so Maximum value can not be find..!!"
        
    def min(self):
        x = 0
        min = self.A[0]

        for i in range(self.n):
            if isinstance(self.A[i], int):
                x = x+1
                
        if x == self.n:
            for i in range(self.n):
                if self.A[i] < min:
                    min = self.A[i]
            return min
        else:    
            return "All Value of List are not Integer so Minimum value can not be find..!!"
     
    def sum(self):
        x = 0
        sum = 0
        for i in range(self.n):
            if isinstance(self.A[i], int):
                x = x+1
                
        if x == self.n:
            for i in range(self.n):
                sum = sum + self.A[i]
        else:    
            return "All Value of List are not Integer so Sum of all value can not be find..!!"
        
        return sum    
    
    def extend(self, item):
        for i in item:
            self.append(i)
            
    def __make_array(self,capacity):
        #create C type array(static,referential) with size capacity 
        return (capacity*ctypes.py_object)()
```
